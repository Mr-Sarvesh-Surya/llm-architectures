```python
text="Cat sat on the mat , also the Boy sat on the sofa holding 4 books in hand"
```

```python
vocab=sorted(set(text))
print(vocab)
```

```python
## Dictionary comprehension
idx_to_char={i:c for i,c in enumerate(vocab)} # mapping ids with charachters
print(idx_to_char)
char_to_idx={c:i for i,c in enumerate(vocab)}
print("charcter to id",char_to_idx)
```

```python
def encode(s):
    return [char_to_idx[c] for c in s]

print("encoded=",encode("sofa"))

def decode(s):
    return "" .join([idx_to_char[i] for i in s])

print(decode([18, 17, 9, 5]))
```
